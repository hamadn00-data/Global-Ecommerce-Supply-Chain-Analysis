## Data Cleaning

### Check Inconsistent Formating

```python
for name, df in {
    'customers' : customers,
    'inventory' : inventory,
    'marketing_spend' : marketing_spend,
    'price_history' : price_history,
    'products' : products,
    'returns' : returns,
    'suppliers_cost' : suppliers_cost,
    'transactions' : transactions
}.items():
    print(f"\nDataset: {name.upper()}")
    print('-'*30)

    string_cols = df.select_dtypes(include=['object', 'string'])

    for col in string_cols:
      if 'id' in col.lower():
       continue
      if 'date' in col.lower():
        continue
      print(f"\n{col}")
      print(df[col].unique())
```

### Handling Outliers

```python
for name, df in {
    'customers' : customers,
    'inventory' : inventory,
    'marketing_spend' : marketing_spend,
    'price_history' : price_history,
    'products' : products,
    'returns' : returns,
    'suppliers_cost' : suppliers_cost,
    'transactions' : transactions
}.items():
    numeric_cols = df.select_dtypes(include=np.number).columns

    for col in numeric_cols:
        plt.figure(figsize=(8, 4))
        plt.boxplot(df[col].dropna())
        plt.title(f"{name} - {col}")
        plt.ylabel(col)
        plt.show()
```

```python
for name, df in {
    'customers' : customers,
    'inventory' : inventory,
    'marketing_spend' : marketing_spend,
    'price_history' : price_history,
    'products' : products,
    'returns' : returns,
    'suppliers_cost' : suppliers_cost,
    'transactions' : transactions
}.items():
 numeric_cols = df.select_dtypes(include=np.number).columns
 for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower)| (df[col] > upper)]
    print(f"\n{name}")
    print(f"{col} - {len(outliers)} outliers")
          
```
